# Download VCT Source Data

Download the latest public version of the [Valorant Champions Tour 2021-2023 dataset](https://www.kaggle.com/datasets/ryanluong1/valorant-champion-tour-2021-2023-data) with KaggleHub, then copy it to the project data directory expected by the analysis notebooks.

Run this notebook with the `vct_data` project directory as the working directory. KaggleHub may require authentication depending on your environment. Keep credentials outside this repository and never commit them.

```python
from pathlib import Path
import shutil
import kagglehub
```

```python
dataset_slug = "ryanluong1/valorant-champion-tour-2021-2023-data"
download_path = Path(kagglehub.dataset_download(dataset_slug))
print("Dataset cache path:", download_path)
```

```python
target_dir = Path("kaggle") / "input" / "valorant-champion-tour-2021-2023-data"
target_dir.mkdir(parents=True, exist_ok=True)
shutil.copytree(download_path, target_dir, dirs_exist_ok=True)
print("Dataset files copied to:", target_dir.resolve())
```
